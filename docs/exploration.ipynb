```python
import requests
import pandas as pd
from datetime import datetime, timedelta, timezone

URL = "https://earthquake.usgs.gov/fdsnws/event/1/query"
params = {
    "format": "geojson",
    "starttime": (datetime.now(timezone.utc) - timedelta(days=30)).strftime("%Y-%m-%d"),
    "latitude": -33.45,      # Santiago
    "longitude": -70.66,
    "maxradiuskm": 300,      # radio alrededor del punto
    "orderby": "time",
    "limit": 20,
}

r = requests.get(URL, params=params, timeout=30)
r.raise_for_status()
features = r.json()["features"]
print(f"{len(features)} sismos encontrados")

rows = []
for f in features:
    p = f["properties"]
    lon, lat, depth = f["geometry"]["coordinates"]
    rows.append({
        "event_id": f["id"],
        "time_utc": pd.to_datetime(p["time"], unit="ms", utc=True),
        "mag": p["mag"],
        "place": p["place"],
        "depth_km": depth,
        "lat": lat,
        "lon": lon,
    })

df = pd.DataFrame(rows)
if not df.empty:
    df["time_local"] = df["time_utc"].dt.tz_convert("America/Santiago")
df
```
